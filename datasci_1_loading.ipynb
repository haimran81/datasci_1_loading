## Excel (XLSX) Datasets

```python
excel1_df = pd.read_excel("data/dataset1.xlsx")

excel1_df.head(10)
```

#### Excel Dataset 1 Features

- **OrderDate** – The date when the customer order was placed.
- **UserId** – A unique identifier for the user or customer who placed the order.
- **ProductId** – A unique identifier for the product that was ordered.
- **PinCode** – The postal or ZIP code associated with the order or customer location.
- **Revenue** – The amount of revenue generated from the order.
- **Units** – The number of product units ordered.
- **Cancelled_Units** – The number of units from the order that were cancelled.

#### Excel Dataset 2 Features

- **Name** – The name of the customer who made the transaction.
- **Transaction_date** – The date and time when the transaction occurred.

```python
excel3_df = pd.read_excel("data/dataset3.xlsx")

excel3_df.head(10)
```

#### Excel Dataset 3 Features

- **Row ID** – A unique number assigned to each row or transaction record.
- **Order ID** – A unique identifier assigned to each customer order.
- **Order Date** – The date when the customer placed the order.
- **Ship Date** – The date when the order was shipped.
- **Ship Mode** – The shipping method used for the order, such as First Class, Second Class, Standard Class, or Same Day.
- **Customer ID** – A unique identifier assigned to each customer.
- **Customer Name** – The name of the customer who placed the order.
- **Segment** – The customer category, such as Consumer, Corporate, or Home Office.
- **Postal Code** – The postal or ZIP code associated with the customer's location.
- **City** – The city associated with the order.
- **State** – The state or administrative area associated with the order.
- **Country** – The country where the order was placed or delivered.
- **Region** – The geographic region associated with the order.
- **Market** – The broader geographic market associated with the order.
- **Product ID** – A unique identifier assigned to the product.
- **Category** – The main category of the product, such as Technology or Furniture.
- **Sub-Category** – A more specific classification of the product, such as Phones, Chairs, or Copiers.
- **Product Name** – The full name or description of the product purchased.
- **Sales** – The total sales amount generated by the transaction.
- **Quantity** – The number of units purchased in the transaction.
- **Discount** – The discount applied to the transaction.
- **Profit** – The amount of profit earned from the transaction.
- **Shipping Cost** – The cost associated with shipping the order.
- **Order Priority** – The priority level assigned to the order, such as Critical, High, or Medium.
